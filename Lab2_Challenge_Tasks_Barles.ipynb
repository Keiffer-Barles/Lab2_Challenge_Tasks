```python
SURNAME = "Barles"
SEED_NUM = 8

# Exercise 1: Sensor Readings Validator
raw_readings = [f"{SURNAME}_{SEED_NUM}", "45.5", "INVALID", f"{SEED_NUM * 10}", "-999", "12.3"]
valid_readings = []

print("=== ASSESSMENT DATA ===")
print("Generated Sensor Data:", raw_readings)

print("\n--- EXECUTION LOG ---")
for idx, entry in enumerate(raw_readings, 1):
    try:
        val = float(entry)
        valid_readings.append(val)
        print(f"[LOG {idx}] Processing '{entry}': Successfully converted to float ({val}). Status: VALID")
    except ValueError:
        print(f"[LOG {idx}] Processing '{entry}': Failed conversion. Status: INVALID (Skipped)")

print("\n=== FINAL OUTPUT ===")
print("Valid Readings:", valid_readings)
if valid_readings:
    print(f"Summary -> Min: {min(valid_readings)}, Max: {max(valid_readings)}, Average: {sum(valid_readings)/len(valid_readings):.2f}")
```

```python
# Exercise 2: Signal Processing & Character Analysis
signal = f"{SURNAME}#{SEED_NUM}!"
letters = digits = special = 0

print("=== ASSESSMENT DATA ===")
print("Generated Signal:", signal)

print("\n--- EXECUTION LOG ---")
for idx, char in enumerate(signal, 1):
    if char.isalpha():
        letters += 1
        char_type = "Letter"
    elif char.isdigit():
        digits += 1
        char_type = "Digit"
    else:
        special += 1
        char_type = "Special Character"
    print(f"[LOG {idx}] Analyzing char '{char}': Identified as {char_type}")

if letters > digits:
    classification = "TEXT DOMINANT"
elif digits > letters:
    classification = "NUMERIC DOMINANT"
else:
    classification = "HYBRID SIGNAL"

print("\n=== FINAL OUTPUT ===")
print(f"Character Analysis -> Letters: {letters}, Digits: {digits}, Special: {special}")
print("Signal Classification:", classification)
```

```python
# Exercise 3: Electronic System Authentication
correct_password = f"{SURNAME}_{SEED_NUM}"
attempt_limit = SEED_NUM if SEED_NUM > 0 else 3
attempts_made = 0
access_granted = False

simulated_inputs = ["wrong_pass", "1234", f"{SURNAME}_{SEED_NUM}"]

print("=== ASSESSMENT DATA ===")
print("Generated Password:", correct_password)
print("Attempt Limit:", attempt_limit)

print("\n--- EXECUTION LOG ---")
for pwd_attempt in simulated_inputs:
    attempts_made += 1
    print(f"[LOG Attempt {attempts_made}/{attempt_limit}] User input: '{pwd_attempt}'")
    
    if pwd_attempt == correct_password:
        access_granted = True
        print("[LOG System] Password matches. Access Granted.")
        break
    else:
        print("[LOG System] Password mismatch. Access Denied.")
        
    if attempts_made >= attempt_limit:
        print("[LOG System] Maximum attempt limit reached. Terminating authentication.")
        break

print("\n=== FINAL OUTPUT ===")
print("Attempts Made:", attempts_made)
print("Access Result:", "SUCCESS" if access_granted else "FAILED")
print("Final System State:", "UNLOCKED" if access_granted else "LOCKED")
```
